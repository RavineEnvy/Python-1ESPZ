```python
#Ex. 1 - Pega o nome e três notas de uma aluno via inpu(), e guarde um dicionário, calcule a média e dusa if para imprimir "Arpovado e "Reprovado.

dicionario = {
    'nome': input('Digite o nome do aluno: '),
    'nota1': float(input('Digite a primeira nota: ')),
    'nota2': float(input('Digite a segunda nota: ')),
    'nota3': float(input('Digite a terceira nota: '))
}

media = (dicionario['nota1'] + dicionario['nota2'] + dicionario['nota3']) / 3

if media >= 7:
    print(f"{dicionario['nome']} está aprovado com média {media:.2f}")
else:
    print(f"{dicionario['nome']} está reprovado com média {media:.2f}")
```

```python
dic = {}

dic['nome'] = input('Digite seu nome: ')

for i in range(3):
    nota = float(input(f'Digite a nota {i+1}: '))
    dic[f'nota{i+1}'] = nota

media = (dic['nota1'] + dic['nota2'] + dic['nota3'])/3

if media >= 7:
    print(f"O aluno(a) {dic['nome']} foi aprovado com a média de {media:.1f}!")
else:
    print(f"O aluno(a) {dic['nome']} foi reprovado com a média de {media:.1f}!")

```

```python
#Ex. 3 - Crie um dicionário livro com páginas_lidas e total_paginas, calcule o percentual lido e classifique com IF

livro = {
    "lidas": int(input("Quantas páginas você leu?")),
    "totais": 520
}

porcentagem = (livro["lidas"]/livro["totais"]) * 100

if livro['totais'] >= livro["lidas"]:
    print(f"Você leu {porcentagem:.1f}% do livro")
else:
    print("Você ultrapassou o número de páginas! Digite um valor válido.")
```

```python
#Ex. 1 - Crie um algorimo utilizando dicionários a qual possui um menu com 
#1 - cadastrar
#2 - Consultar
#3 - Sair
# opcao = escolha opcao 1, 2 ou 3

#Enquanto opcao nao esta em menu
#   opcao = opcao invalida
#voce escolheu a opcao x do menu

menu = {
    1: "cadastrar",
    2: "consultar",
    3: "sair"
}

opcao = int(input("Digite a sua opção: "))

while opcao not in menu:
    opcao = int(input("Digite a sua opção: "))

print(f"opção escolhida {menu[opcao]}")
```
