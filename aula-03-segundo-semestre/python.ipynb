```python
class Pessoa:
    def __init__(self,nome,idade):
        self.nome = nome
        self.idade = idade
    def apresentar(self):
        if self.idade > 1:
            return f"Olá, você é {self.nome},e está com {self.idade} anos."
        else:
            return f"Olá, você é {self.nome}, e está com {self.idade} ano."

p1= Pessoa("Ana", 15)
p2= Pessoa("Guilherme", 26)
p3= Pessoa("Arthur", 1)

print(p1.apresentar())
print(p2.apresentar())
print(p3.apresentar())
```

```python
#Ex. 2 - calculo da área e perímero de um retângulo

class Retangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    def area(self):
        return self.base * self.altura
    def perimetro(self):
        return 2 * (self.base + self.altura)

r1 = Retangulo(float(input("Base: ")), float(input("Altura: ")))

print(f"A Área do r1: {r1.area()} - O Perímetro do r1: {r1.perimetro()}")
print(r1.__dict__)
```

```python
import pandas as pd
#Ex - Crie um algoritmo utilizando class para calcular o índice de massa corporal para 3 pessoas, Variáveis de entrada: mome, peso, altura

#Cálculo peso / altura**2

# Criar 3 objetos (nome, peso, altura)

#testar para os 3 objetos

class MassaCorporal:
    def __init__(self, nome, peso, altura):
        self.nome = nome
        self.peso = peso
        self.altura = altura
    def calcularMassa(self):
        return self.peso / (self.altura**2)
    def exibir(self):
        return f"{self.nome} tem {self.peso}Kg - {self.altura}M - Massa de {self.calcularMassa():.2f}"

lista = []

for i in range (1,4):
    i = MassaCorporal(str(input("Nome: ")), float(input("Peso: ")), float(input("Altura (em Metros): ")))
    lista.append(i)
    print(i.exibir())

dados = [
    {'nome':i.nome, "Peso":i.peso, 'Altura': i.altura, 'IMC': i.calcularMassa()}
    for i in lista
]

df = pd.DataFrame(dados)
print(df)
```

```python
#Ex - Calcular juros compostos de 3 investimentos (CDB, Tesouro Selic, Fundo Y), tendo as seguintes variáveis: nome, capital, taxa (taxa em % ao mês) e tempo (meses)

#Montante = capital * (1 + taxa/ 100) **tempo
#Lucro - montante - capital

#Criar 3 objetos

class JurosCompostos:
    def __init__(self, nome, capital, taxa, tempo):
        self.nome = nome
        self.capital = capital
        self.taxa = taxa
        self.tempo = tempo
    def calcularJuros(self):
        return self.capital * (1 + self.taxa / 100) ** self.tempo
    def lucro(self):
        return self.calcularJuros() - self.capital
    def exibirResultado(self):
        return f"Olá, {self.nome}! Lucro: R${self.lucro():.2f} - Total com Juros: R${self.calcularJuros():.2f}"

for i in range (1,4):
    i = JurosCompostos(str(input("Nome: ")), float(input("Capital: ")), float(input("Taxa(em %): ")), int(input("Tempo(em meses): ")))
    print(i.exibirResultado())
```
