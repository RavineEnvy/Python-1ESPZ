```python
class NomedaClasse:
    def __init__(self, atributo1, atributo2):
        self.atributos1 = atributo1
        self.atributos2 =  atributo2

    def metodo(self):
        print(self.atributos1)
        print(self.atributos2)

a1 = NomedaClasse("A", "B")
a1.metodo()
```

```python
class Pessoa:
    def __init__(self,nome,idade):
        self.nome = nome
        self.idade = idade
    def apresentar(self):
        if self.idade > 1:
            return f"Olá, você é {self.nome},e está com {self.idade} anos."
        else:
            return f"Olá, você é {self.nome}, e está com {self.idade} ano."

p1= Pessoa("Ana", 15)
p2= Pessoa("Guilherme", 26)
p3= Pessoa("Arthur", 1)

print(p1.apresentar())
print(p2.apresentar())
print(p3.apresentar())
```

```python
class Bolo:
    def __init__(self, sabor):
        self.sabor = sabor

b1 = Bolo("Chocolate")
b2 = Bolo("Morango")

print(f"Bolo 1 tem o sabor - {b1.sabor}. E o bolo 2 tem o sabor - {b2.sabor}")
```

```python
#ex crie ua classe chamada Estudante, a qual apresente quatro variáveis:
#nome, faculdade, curso, semestre.

#Após isso, apresente um método para utilizar e mostrar as quatro variáveis e mostrar um texto usando todas as variáveis

class Estudante:
    def __init__(self, nome,faculdade, curso, semestre):
        self.nome = nome
        self.faculdade = faculdade
        self.curso = curso
        self.semestre = semestre
    def apresentar(self):
        print(f"O aluno {self.nome}, está na faculdade {self.faculdade}, no curso de {self.curso} e no {self.semestre}º semestre.")

a1 = Estudante("Guilherme","FIAP","Engenharia de Software", 2)

a1.apresentar()
```
