```python
import pandas as pd

#Ex. 1 - Crie um algoritmo para calcular o desconto e mostrar o respectivo salário de uma determinada pessoa.
#Insira nome, salario, taxa de desconto (27.5%) e retorne o valor de desconto e o valor a ser recebido
#Faça o teste para duas ou três pessoas

class Desconto:
    def __init__(self, nome, salario, desconto):
        self.nome = nome
        self.salario = salario
        self.desconto = desconto
    def calcularDesconto(self):
        descontar = self.salario * (self.desconto/100)
        return descontar
    def valorFinal(self):
        valorFinal = self.salario - self.calcularDesconto()
        return valorFinal
    def apresentar(self):
        return f"Nome: {self.nome} - Salário: R${self.salario} - Desconto: R${self.calcularDesconto():.2f} - Valor final: R${self.valorFinal():.2f}.)"

# p1 = Desconto(input('Digite seu nome: '), float(input("Digite seu salário: ")))
# p2 = Desconto(input('Digite seu nome: '), float(input("Digite seu salário: ")))

# print(p1.apresentar())
# print(p2.apresentar())

#Ex. 2 - para o exercício anterior, crie mais 4 objetos, crie uma lista de objetos, depois utilize uma estrutura de repetição para mostrar um texto com o respectivo desconto e salário a receber para cada pessoa.

lista = []
listaTexto = []
listaDict = []

for i in range (4):
    i = Desconto(input('Digite seu nome: '), float(input("Digite seu salário: ")), float(input("Digite a porcentagem do Desconto: ")))
    lista.append(i)
    listaTexto.append(i.apresentar())

    linha = {
        "Nome" : i.nome,
        "Salário" : i.salario,
        "Desconto em %" :i.desconto,
        "Valor Descontado" : i.calcularDesconto(),
        "Valor Final" : i.valorFinal()
    }

    listaDict.append(linha)

print(listaTexto)

print(pd.DataFrame(listaDict))

#Outra maneira de criar um Data Frame

dados = [{
        "Nome" : i.nome,
        "Salário" : i.salario,
        "Desconto em %" :i.desconto,
        "Valor Descontado" : i.calcularDesconto(),
        "Valor Final" : i.valorFinal(),
    } for i in lista]

print(pd.DataFrame(dados))

df = pd.DataFrame(dados)

df.to_excel('lista_dados.xlsx')
df.to_csv('lista_dadosCSV.csv')
```

```python
from matplotlib import pyplot as plt
import pandas as pd

df1 = pd.read_csv('manutencao_preditiva(in).csv')

x = df1['UDI']
y1 = df1['Temperatura Ar [K]']
y2 = df1['Temperatura Processo [K]']

# Cria a figura e os eixos
plt.figure(figsize=(10, 6))

# Plota a Temperatura do Ar
plt.plot(x, y1, label='Temperatura Ar [K]', color='blue', alpha=0.7)

# Plota a Temperatura do Processo
plt.plot(x, y2, label='Temperatura Processo [K]', color='red', alpha=0.7)

# Adiciona título e rótulos nos eixos
plt.title('Comparação: Temperatura do Ar vs Temperatura do Processo', fontsize=14)
plt.xlabel('UDI (Identificador)', fontsize=12)
plt.ylabel('Temperatura [K]', fontsize=12)



# Mostra a legenda (usa os labels definidos no plt.plot)
plt.legend()

# Ajusta o layout para evitar cortes
plt.tight_layout()

# Exibe o gráfico
plt.show()

```
