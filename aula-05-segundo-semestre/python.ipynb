```python
#Ex. 1
def soma_ate(n):
    total = 0
    for i in range(1, n + 1):
        total= total + i
    return total

#Ex. 2
def soma_ate2(n):
    return n * (n + 1) // 2


%time print(soma_ate(6))
%time print(soma_ate2(6))
```
